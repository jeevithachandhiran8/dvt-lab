```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create Dataset
data = {
    "Roll_No": [201, 202, 203, 204, 205, 206, 207, 208],
    "Name": ["Aakash", "Meena", "Sanjay", "Kavya", "Rahul", "Nisha", "Varun", "Sneha"],
    "Department": ["CSE", "IT", "AI&DS", "ECE", "CSE", "IT", "AI&DS", "ECE"],
    "Marks": [82, 76, 91, 88, 94, 81, 86, 79],
    "Attendance": [90, 85, 95, 92, 97, 89, 91, 87]
}

# Convert dictionary into DataFrame
df = pd.DataFrame(data)

# Display Dataset
print(df)
```

```python
print(df.shape)
```

```python
df.describe()
```

```python
high_marks = df[df["Marks"] > 85]
print(high_marks)
```

```python
cse_students = df[df["Department"] == "CSE"]
print(cse_students)
```

```python
sorted_df = df.sort_values(by="Marks", ascending=False)
print(sorted_df)
```

```python
plt.figure(figsize=(8,5))
plt.bar(sorted_df["Name"], sorted_df["Marks"])
plt.title("Student Marks Ranking")
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.show()
```

```python

#Statistical Analysis
print("Mean:", df["Marks"].mean())
print("Median:", df["Marks"].median())
print("Mode:")
print(df["Marks"].mode())
print("Standard Deviation:", df["Marks"].std())
```

```python
dept_avg = df.groupby("Department")["Marks"].mean()
print(dept_avg)
```

```python
dept_avg.plot(kind="bar")
plt.title("Department-wise Average Marks")
plt.xlabel("Department")
plt.ylabel("Average Marks")
plt.show()
```

```python
correlation = df[["Marks", "Attendance"]].corr()
print(correlation)
```
