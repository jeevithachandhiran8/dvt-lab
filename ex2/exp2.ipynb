```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("student.db")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    marks INTEGER
)
""")

cursor.execute("DELETE FROM students")

cursor.execute("INSERT INTO students VALUES (1, 'Alice', 20, 90)")
cursor.execute("INSERT INTO students VALUES (2, 'Bob', 21, 85)")
cursor.execute("INSERT INTO students VALUES (3, 'Charlie', 19, 95)")

conn.commit()

df = pd.read_sql("SELECT * FROM students", conn)
print(df)

conn.close()
```
