```python
import pandas as pd

df=pd.read_csv("C:/Users/91754/Downloads/transactions.csv")
print(df.head())
```

```python
print(df.isnull())
```

```python
print(df.isnull().sum())
```

```python
df["Amount (INR)"] = df["Amount (INR)"].fillna(df["Amount (INR)"].mean())

df["Sender Name"] = df["Sender Name"].fillna("Unknown")
df["Sender UPI ID"] = df["Sender UPI ID"].fillna("Unknown")
df["Receiver Name"] = df["Receiver Name"].fillna("Unknown")
df["Receiver UPI ID"] = df["Receiver UPI ID"].fillna("Unknown")
df["Status"] = df["Status"].fillna("Unknown")

print(df)
```

```python
print("Duplicate Records:")

print(df.duplicated())

print("Total Duplicates:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Dataset after removing duplicates")

print(df)
```

```python
Q1 = df["Amount (INR)"].quantile(0.25)

Q3 = df["Amount (INR)"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR

upper = Q3 + 1.5 * IQR

outliers = df[(df["Amount (INR)"] < lower) |
              (df["Amount (INR)"] > upper)]

print(outliers)
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()

df[["Amount (INR)"]] = scaler.fit_transform(df[["Amount (INR)"]])

print(df[["Amount (INR)"]])
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df["Status"] = encoder.fit_transform(df["Status"])

print(df)
```
